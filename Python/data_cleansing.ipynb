```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("../data/data.csv")
data.head()
# Hay 26 variables
```

```python
data.info()
```

```python
data.nunique()
# Notamos que solamente hay 14 variables continuas de a cuerdo a la documentación: age, bp, bgr, bu, sc, sod, pot, hemo, pcv, wc, rc. Nos quedaremos solamente con esas.
# Véase documentación en: https://archive.ics.uci.edu/dataset/336/chronic+kidney+disease
```

```python
data = data.drop(columns=["id", "rbc", "pc", "pcc", "ba", "htn",
                          "dm", "cad", "appet", "pe", "ane", "classification","su","sg","al"])
data.isna().sum()
```

```python
# Además, por recomendación de nuestro primer entrevistado, Max Quintero, rechazamos las variables "rc" y "pcv", ya que reporta que ambas, junto con "hemo", miden lo mismo, y "hemo" es la que
# menos faltantes de informacion tiene.
data=data.drop(columns=["pcv","rc"])
data=data.dropna()
data.info()
```

```python
# Podemos observar que todas las variables son claramente continuas, salvo una: bp. En la documentacion del dataset no se menciona si esta es presión sistólica o diastólica. En base a la media
# de los valores (alrededor de 80), podemos afirmar que se trata de presión diastólica. Entonces, hay dos valores que son claramente outliers que probablemente representen
# errores de captura; así que decidimos eliminarlos. Con eso, nos quedan tan solo 7 valores que la variable aleatoria puede tomar. A pesar de contar con tan pocos valores posibles (dificilmente continua, 
# decidimos conservarla en el dataset final puesto que las personas que entrevistamos le dieron mucha importancia a esta variable. 
```

```python
data = data[~data['bp'].isin([140, 180])]
```

```python
# Notemos que incluso quitando dos de las columnas con más NaNs el dataset limpio crece de apenas 203 registros completos a 219 (NOTA: si se corre el data.dropna() con el dataset SIN variables 
# categóricas, es decir, antes de eliminar las otras dos variables por recomendación del estudiante de medicina, hay solamente 219 registros completos). Esto se puede interpretar como que los faltantes
# están bien dispersos a lo largo de las variables, por lo que imputar datos no es opción ya que se estarían estimando perfiles médicos. Decidimos pues, quedarnos con este ultimo dataset.
```

```python
data.to_csv('../data/data_clean.csv', index=False)
```
